```python
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
import cv2
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from collections import Counter

# --- CÁC IMPORT TỪ PROJECT CỦA BẠN ---
# Đảm bảo các đường dẫn này đúng với cấu trúc project trên Kaggle
import config
from utils import load_trained_model # Giả định hàm này load model từ file .h5 hoặc .keras
from data_operations.data_preprocessing import load_inbreast_data_no_pectoral_removal

# =================================================================
# BƯỚC 1: CẤU HÌNH VÀ TẢI LẠI MÔ HÌNH ĐÃ HUẤN LUYỆN
# =================================================================
print("--- 1. Tải lại mô hình đã huấn luyện ---")

# --- Đặt các thông số để tải đúng model ---
# Tên của lần chạy tốt nhất
# LƯU Ý: Hãy thay thế bằng tên chính xác của file model bạn đã lưu
# Ví dụ: 'INbreast_ROI_SGD_BalancedReg' hoặc 'INbreast_ROI_SGD_BalancedReg.keras'
MODEL_NAME = 'INbreast_ROI_SuperAug' 
MODEL_PATH = f'/kaggle/input/superaug/tensorflow2/default/1/dataset-INbreast_type-all_model-MobileNet_lr-0.0005_b-16_e1-40_e2-80_roi-True_INbreast_ROI_SuperAug.weights.h5'
mock_args.cli_args = mock_args # Cần cho lỗi 'cli_args is not defined'

# Tải lại model
# Lưu ý: Cần có file kiến trúc json nếu bạn lưu riêng
try:
    # Nếu bạn có file kiến trúc json:
    # with open(f'/path/to/{MODEL_NAME}_arch.json', 'r') as json_file:
    #     loaded_model_json = json_file.read()
    # model = tf.keras.models.model_from_json(loaded_model_json)
    # model.load_weights(MODEL_PATH)
    
    # Nếu bạn có thể load trực tiếp (ví dụ từ CnnModel class)
    # Đây là cách đơn giản hơn nếu CnnModel có thể khởi tạo lại kiến trúc
    from cnn_models.cnn_model import CnnModel
    cnn = CnnModel("MobileNet", num_classes=2, args=mock_args)
    cnn.model.load_weights(MODEL_PATH)
    model = cnn.model
    model.summary()
    print(f"Tải thành công model từ: {MODEL_PATH}")

except Exception as e:
    print(f"Lỗi khi tải model: {e}")
    # Dừng ở đây nếu không tải được model
    model = None

# =================================================================
# BƯỚC 2: TẢI VÀ CHUẨN BỊ DỮ LIỆU TEST
# =================================================================
if model:
    print("\n--- 2. Chuẩn bị dữ liệu để visualize ---")
    DATA_DIR = '/kaggle/input/breastdata/INbreast/INbreast'
    le = LabelEncoder()
    
    # Tải lại toàn bộ dữ liệu (chúng ta chỉ cần tập test)
    X_loaded, y_text_labels = load_inbreast_data_no_pectoral_removal(
        data_dir=DATA_DIR,
        label_encoder=le,
        use_roi_patches=True, # Chạy với ROI
        target_size=(224, 224)
    )
    
    le.fit(y_text_labels)
    y_numeric_labels = le.transform(y_text_labels)

    # Chia lại y hệt như lúc train để có tập test chính xác
    _, X_test, _, y_test_numeric = train_test_split(
        X_loaded, y_numeric_labels, test_size=0.2, stratify=y_numeric_labels, random_state=42
    )
    print(f"Đã tải và chia xong, có {len(X_test)} ảnh trong tập test.")

# =================================================================
# BƯỚC 3: ĐỊNH NGHĨA HÀM GRAD-CAM
# =================================================================
def make_gradcam_heatmap(img_array, model, last_conv_layer_name, pred_index=None):
    grad_model = tf.keras.models.Model(
        [model.inputs], [model.get_layer(last_conv_layer_name).output, model.output]
    )

    with tf.GradientTape() as tape:
        last_conv_layer_output, preds = grad_model(img_array)
        if pred_index is None:
            pred_index = tf.argmax(preds[0])
        class_channel = preds[:, pred_index]

    grads = tape.gradient(class_channel, last_conv_layer_output)
    pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2))

    last_conv_layer_output = last_conv_layer_output[0]
    heatmap = last_conv_layer_output @ pooled_grads[..., tf.newaxis]
    heatmap = tf.squeeze(heatmap)
    heatmap = tf.maximum(heatmap, 0) / tf.math.reduce_max(heatmap)
    return heatmap.numpy()

def display_gradcam(img, heatmap, alpha=0.4):
    # Chuyển ảnh về dạng uint8 để hiển thị
    img_u8 = (img * 255).astype(np.uint8)
    
    # Resize heatmap về kích thước ảnh gốc
    heatmap = cv2.resize(heatmap, (img.shape[1], img.shape[0]))
    heatmap = np.uint8(255 * heatmap)
    heatmap = cv2.applyColorMap(heatmap, cv2.COLORMAP_JET)

    # Trộn heatmap và ảnh gốc
    superimposed_img = cv2.addWeighted(cv2.cvtColor(img_u8, cv2.COLOR_RGB2BGR), 1-alpha, heatmap, alpha, 0)
    
    return cv2.cvtColor(superimposed_img, cv2.COLOR_BGR2RGB)


# =================================================================
# BƯỚC 4: TẠO VÀ HIỂN THỊ VISUALIZATION
# =================================================================
if model and 'X_test' in locals():
    print("\n--- 4. Tạo và hiển thị Grad-CAM ---")
    
    # Tên của lớp conv cuối cùng trong MobileNetV2
    # Bạn có thể tìm tên này bằng cách `model.summary()`
    LAST_CONV_LAYER_NAME = "out_relu" 

    # Chọn một vài ảnh từ tập test để visualize
    # Ví dụ: 2 ảnh Benign và 2 ảnh Malignant
    benign_indices = np.where(y_test_numeric == 0)[0]
    malignant_indices = np.where(y_test_numeric == 1)[0]
    
    sample_indices = np.concatenate([
        np.random.choice(benign_indices, 2, replace=False),
        np.random.choice(malignant_indices, 2, replace=False)
    ])

    plt.figure(figsize=(12, 12))
    
    for i, img_idx in enumerate(sample_indices):
        original_img = X_test[img_idx]
        img_array = np.expand_dims(original_img, axis=0) # Thêm chiều batch

        # Dự đoán để lấy nhãn
        preds = model.predict(img_array)
        predicted_class_idx = np.argmax(preds[0])
        predicted_class_name = le.classes_[predicted_class_idx]
        true_class_name = le.classes_[y_test_numeric[img_idx]]

        # Tạo heatmap
        heatmap = make_gradcam_heatmap(img_array, model, LAST_CONV_LAYER_NAME)
        
        # Overlay heatmap lên ảnh
        superimposed_img = display_gradcam(original_img, heatmap, alpha=0.5)

        # Hiển thị
        plt.subplot(4, 2, 2*i + 1)
        plt.imshow(original_img)
        plt.title(f"Original - True: {true_class_name}")
        plt.axis('off')

        plt.subplot(4, 2, 2*i + 2)
        plt.imshow(superimposed_img)
        plt.title(f"Grad-CAM - Predicted: {predicted_class_name}")
        plt.axis('off')

    plt.tight_layout()
    plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

# Định nghĩa các cấp độ pyramid
levels = [1, 2, 3, 4, 5]
x = levels
y_c = 0  # backbone feature maps (C) ở y=0
y_p = 1  # pyramid outputs (P) ở y=1

# Tạo danh sách mũi tên lần lượt hiện thị
arrow_defs = []

# 1) FPN top-down flow (P_{i+1} → P_i)
for i in range(len(levels) - 1):
    start = (x[i + 1], y_p)
    end   = (x[i],     y_p)
    arrow_defs.append((start, end))

# 2) RevFP bottom-up flow (P_{i+1} → C_i)
for i in range(len(levels) - 1):
    start = (x[i + 1], y_p)
    end   = (x[i],     y_c)
    arrow_defs.append((start, end))

# 3) RevFP local top-down (C_{i+1} → P_i)
for i in range(len(levels) - 1):
    start = (x[i + 1], y_c)
    end   = (x[i],     y_p)
    arrow_defs.append((start, end))

# Thiết lập figure
fig, ax = plt.subplots(figsize=(8, 3))
ax.set_xlim(min(x) - 1, max(x) + 1)
ax.set_ylim(-0.5, 1.5)
ax.axis('off')

# Hàm vẽ các nút C_i và P_i
def draw_nodes():
    for xi, lvl in zip(x, levels):
        ax.scatter(xi, y_c, s=200, edgecolors='black')
        ax.text(xi, y_c, f"C{lvl}", ha='center', va='center')
        ax.scatter(xi, y_p, s=200, edgecolors='black')
        ax.text(xi, y_p, f"P{lvl}", ha='center', va='center')

# Hàm cập nhật cho mỗi frame
def update(frame):
    ax.clear()
    ax.axis('off')
    draw_nodes()
    # Vẽ tất cả các mũi tên up đến frame hiện tại
    for idx in range(frame + 1):
        start, end = arrow_defs[idx]
        ax.annotate(
            '', xy=end, xytext=start,
            arrowprops=dict(arrowstyle='->', lw=2)
        )
    return []

# Tạo animation: tổng số frames = len(arrow_defs)
anim = FuncAnimation(
    fig, update,
    frames=len(arrow_defs),
    interval=500,  # mili-giây giữa các frame
    repeat=True
)

# Nếu chạy trong Jupyter:
from IPython.display import HTML
display(HTML(anim.to_jshtml()))

# Nếu chạy như script .py, thay 2 dòng trên bằng:
# plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from matplotlib.animation import FuncAnimation
from matplotlib.patches import Rectangle
from IPython.display import HTML

# --- 1) Load and downsample mammogram ---
img = Image.open('/home/neeyuhuynh/Desktop/Breast-Cancer-Detection-Mammogram-Deep-Learning-Publication/data/CMMD/benign/D2-0092_1-1.png').convert('L')
input_img = np.array(img)
# Downsample to make animation lighter
input_img = input_img[::4, ::4]

# --- 2) Define kernel ---
kernel = np.array([[1, 0, -1],
                   [1, 0, -1],
                   [1, 0, -1]])
kH, kW = kernel.shape
H, W = input_img.shape
out_h, out_w = H - kH + 1, W - kW + 1

# --- 3) Precompute min/max for normalization ---
all_vals = [(input_img[i:i+kH, j:j+kW] * kernel).sum()
            for i in range(out_h) for j in range(out_w)]
abs_max = max(abs(min(all_vals)), abs(max(all_vals)))

# --- 4) Set up figure and images ---
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(6, 2.5))
for ax in (ax1, ax2, ax3):
    ax.set_xticks([]); ax.set_yticks([])

# Input + patch rectangle
im1 = ax1.imshow(input_img, cmap='gray', vmin=0, vmax=255)
rect = Rectangle((0, 0), kW, kH, edgecolor='red', fill=False, linewidth=1.5)
ax1.add_patch(rect)
ax1.set_title('Input Patch')

# Patch * kernel
im2 = ax2.imshow(np.zeros((kH, kW)), cmap='bwr', vmin=-abs_max, vmax=abs_max)
ax2.set_title('Patch × Kernel')

# Feature map output
output = np.zeros((out_h, out_w))
im3 = ax3.imshow(output, cmap='bwr', vmin=-abs_max, vmax=abs_max)
ax3.set_title('Accumulated Feature Map')
plt.tight_layout()

# --- 5) Animation update ---
def update(frame):
    pos, phase = divmod(frame, 2)
    i, j = divmod(pos, out_w)
    
    # Move patch window
    rect.set_xy((j, i))
    
    if phase == 0:
        # Phase 1: highlight only
        im2.set_data(np.zeros((kH, kW)))
    else:
        # Phase 2: compute and update
        patch = input_img[i:i+kH, j:j+kW]
        prod = patch * kernel
        output[i, j] = prod.sum()
        im2.set_data(prod)
        im3.set_data(output)
    
    return rect, im2, im3

frames = out_h * out_w * 2
anim = FuncAnimation(fig, update, frames=frames, interval=200, blit=True)

# --- 6) Display as HTML5 video ---
HTML(anim.to_html5_video())
```

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
import os
import pydicom  # Thư viện để đọc file DICOM

# =================================================================
# --- 1. CONFIGURATION --------------------------------------------
# =================================================================
# ✅ THAY ĐỔI CÁC ĐƯỜNG DẪN DƯỚI ĐÂY

# --- BƯỚC 1: CHỈ ĐỊNH THƯ MỤC NGUỒN (DICOM) VÀ ĐÍCH (PNG) ---

# Thư mục chứa các file DICOM của bạn
DICOM_SOURCE_DIR = '/home/neeyuhuynh/Desktop/Breast-Cancer-Detection-Mammogram-Deep-Learning-Publication/data/INbreast/AllDICOMs' 

# Thư mục để lưu các file PNG sau khi chuyển đổi
# Thư mục này sẽ được tự động tạo nếu chưa có
PNG_OUTPUT_DIR = '/home/neeyuhuynh/Desktop/Breast-Cancer-Detection-Mammogram-Deep-Learning-Publication/data/INbreast/AllPNGs/'

# --- BƯỚC 2: CHỌN TÊN FILE (KHÔNG CÓ ĐUÔI FILE) ĐỂ TRỰC QUAN HÓA ---

# Tên file chính (không có .dcm)
FILENAME_1 = '22678622_69a72d3529577d61_MG_R_CC_ANON'

# Tên file thứ hai để dùng cho CutMix và Mixup (không có .dcm)
FILENAME_2 = '22678646_69a72d3529577d61_MG_R_MLO_ANON'


# =================================================================
# --- 2. PREPROCESSING FUNCTION (DICOM to PNG) --------------------
# =================================================================

def convert_dicom_to_png(dicom_dir, png_dir):
    """
    Reads all DICOM files from a source directory, converts them to PNG,
    and saves them to a destination directory.
    This version uses force=True to handle non-compliant DICOM files.
    """
    # Create the destination directory if it doesn't exist
    os.makedirs(png_dir, exist_ok=True)
    
    print(f"Starting DICOM to PNG conversion...")
    print(f"Source: {dicom_dir}")
    print(f"Destination: {png_dir}")

    dicom_files = [f for f in os.listdir(dicom_dir) if f.endswith('.dcm')]
    
    converted_count = 0
    failed_count = 0

    for filename in dicom_files:
        dicom_path = os.path.join(dicom_dir, filename)
        png_filename = os.path.splitext(filename)[0] + '.png'
        png_path = os.path.join(png_dir, png_filename)

        # Optional: Skip if the PNG file already exists to save time
        if os.path.exists(png_path):
            continue

        try:
            # ✅ SỬA LỖI: Thêm `force=True` vào đây
            ds = pydicom.dcmread(dicom_path, force=True)
            
            # Get the pixel data
            pixel_array = ds.pixel_array

            # Normalize the pixel values to 0-255 range (for 8-bit PNG)
            pixel_array_float = pixel_array.astype(float)
            
            min_pixel = np.min(pixel_array_float)
            max_pixel = np.max(pixel_array_float)

            if max_pixel - min_pixel == 0:
                normalized_image_8bit = np.zeros(pixel_array_float.shape, dtype=np.uint8)
            else:
                rescaled_image = (pixel_array_float - min_pixel) / (max_pixel - min_pixel)
                normalized_image_8bit = (rescaled_image * 255).astype(np.uint8)

            # Save the image as PNG
            cv2.imwrite(png_path, normalized_image_8bit)
            converted_count += 1
            
        except Exception as e:
            # In ra lỗi nhưng không dừng chương trình
            print(f"Could not convert {filename}. Reason: {e}")
            failed_count += 1
            
    print(f"DICOM to PNG conversion complete. Converted: {converted_count}, Failed: {failed_count}")


# =================================================================
# --- 3. DATA AUGMENTATION FUNCTIONS ------------------------------
# =================================================================

def elastic_transform(image, alpha, sigma, random_state=None):
    if random_state is None:
        random_state = np.random.RandomState(None)
    shape = image.shape
    dx = cv2.GaussianBlur((random_state.rand(*shape[:2]) * 2 - 1), (0,0), sigma) * alpha
    dy = cv2.GaussianBlur((random_state.rand(*shape[:2]) * 2 - 1), (0,0), sigma) * alpha
    x, y = np.meshgrid(np.arange(shape[1]), np.arange(shape[0]))
    map_x = (x + dx).astype('float32')
    map_y = (y + dy).astype('float32')
    return cv2.remap(image, map_x, map_y, interpolation=cv2.INTER_LINEAR, borderMode=cv2.BORDER_REFLECT_101)

def cutmix(image1, image2, lam=0.4):
    h, w, _ = image1.shape
    r_x, r_y = np.random.randint(w), np.random.randint(h)
    r_w, r_h = int(w * np.sqrt(1 - lam)), int(h * np.sqrt(1 - lam))
    x1, y1 = np.clip(r_x - r_w // 2, 0, w), np.clip(r_y - r_h // 2, 0, h)
    x2, y2 = np.clip(r_x + r_w // 2, 0, w), np.clip(r_y + r_h // 2, 0, h)
    mixed_image = image1.copy()
    mixed_image[y1:y2, x1:x2] = image2[y1:y2, x1:x2]
    return mixed_image

def mixup(image1, image2, lam=0.6):
    mixed_image = (lam * image1.astype(np.float32) + (1 - lam) * image2.astype(np.float32))
    return mixed_image.astype(np.uint8)


# =================================================================
# --- 4. LOAD, APPLY, AND VISUALIZE -------------------------------
# =================================================================

def main():
    # --- NEW: Run DICOM to PNG conversion first ---
    convert_dicom_to_png(DICOM_SOURCE_DIR, PNG_OUTPUT_DIR)
    
    # --- Paths now point to the newly created PNG files ---
    IMAGE_PATH_1 = os.path.join(PNG_OUTPUT_DIR, FILENAME_1 + '.png')
    IMAGE_PATH_2 = os.path.join(PNG_OUTPUT_DIR, FILENAME_2 + '.png')

    # Load images from paths and convert to RGB color space
    try:
        img1 = cv2.imread(IMAGE_PATH_1)
        if img1 is None:
            raise FileNotFoundError(f"Could not read image at: {IMAGE_PATH_1}")
        img1 = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)

        img2 = cv2.imread(IMAGE_PATH_2)
        if img2 is None:
            raise FileNotFoundError(f"Could not read image at: {IMAGE_PATH_2}")
        img2 = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)
    except FileNotFoundError as e:
        print(f"Error: {e}")
        print("Please check if the DICOM files exist or if the conversion was successful.")
        return

    # Ensure both images have the same dimensions for Mixup and CutMix
    if img1.shape != img2.shape:
        print("Warning: The two images have different dimensions. Resizing image 2 to match image 1.")
        img2 = cv2.resize(img2, (img1.shape[1], img1.shape[0]))

    # --- Apply augmentation techniques ---
    ELASTIC_ALPHA = 35
    ELASTIC_SIGMA = 4
    CUTMIX_LAMBDA = 0.5
    MIXUP_LAMBDA = 0.6
    
    elastic_img = elastic_transform(img1, alpha=ELASTIC_ALPHA, sigma=ELASTIC_SIGMA)
    cutmix_img = cutmix(img1, img2, lam=CUTMIX_LAMBDA)
    mixup_img = mixup(img1, img2, lam=MIXUP_LAMBDA)

    # --- Visualize the results ---
    fig, axes = plt.subplots(2, 3, figsize=(18, 12))
    fig.suptitle('Visualization of Augmentation Techniques', fontsize=20)

    axes[0, 0].imshow(img1)
    axes[0, 0].set_title('Original Image 1')
    axes[0, 0].axis('off')

    axes[0, 1].imshow(img2)
    axes[0, 1].set_title('Original Image 2')
    axes[0, 1].axis('off')

    axes[0, 2].imshow(elastic_img)
    axes[0, 2].set_title(f'1. Elastic Transformation\n(alpha={ELASTIC_ALPHA}, sigma={ELASTIC_SIGMA})')
    axes[0, 2].axis('off')
    
    axes[1, 0].imshow(cutmix_img)
    axes[1, 0].set_title(f'2. CutMix (with Image 2)\n(lambda={CUTMIX_LAMBDA})')
    axes[1, 0].axis('off')
    
    axes[1, 1].imshow(mixup_img)
    axes[1, 1].set_title(f'3. Mixup (with Image 2)\n(lambda={MIXUP_LAMBDA})')
    axes[1, 1].axis('off')

    axes[1, 2].axis('off')

    plt.tight_layout(rect=[0, 0, 1, 0.96])
    plt.show()


if __name__ == '__main__':
    main()
```
